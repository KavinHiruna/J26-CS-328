```python
from sklearn.ensemble import IsolationForest
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import numpy as np

NORMAL_CLASS = 7


X_train_normal = X_train[y_train == NORMAL_CLASS]

print("Training Isolation Forest...")
start = time.time()
iso_model = IsolationForest(n_estimators=100, contamination=0.1,
                            random_state=42, n_jobs=-1)
iso_model.fit(X_train_normal)
iso_train_time = time.time() - start


start = time.time()
iso_pred_raw = iso_model.predict(X_test)
iso_infer_time = time.time() - start
iso_ms = (iso_infer_time / len(X_test)) * 1000


y_test_bin = (y_test != NORMAL_CLASS).astype(int)
y_pred_iso = (iso_pred_raw == -1).astype(int)

joblib.dump(iso_model, "iso_model.pkl")
iso_size_mb = os.path.getsize("iso_model.pkl") / (1024 * 1024)

print("Isolation Forest Accuracy (binary):", accuracy_score(y_test_bin, y_pred_iso))
print(classification_report(y_test_bin, y_pred_iso, target_names=['Normal', 'Attack']))
print(confusion_matrix(y_test_bin, y_pred_iso))
print(f"ISO train time:        {iso_train_time:.2f} s")
print(f"ISO inference time:    {iso_infer_time:.3f} s for {len(X_test)} samples")
print(f"ISO per-sample latency: {iso_ms:.4f} ms")
print(f"ISO model size:        {iso_size_mb:.2f} MB")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd

NORMAL_CLASS = 7


y_test_bin = (y_test != NORMAL_CLASS).astype(int)   # 0 = Normal, 1 = Attack

rf_bin  = (rf_model.predict(X_test)  != NORMAL_CLASS).astype(int)
xgb_bin = (xgb_model.predict(X_test) != NORMAL_CLASS).astype(int)
iso_bin = (iso_model.predict(X_test) == -1).astype(int)

def row(name, pred, train_t, infer_ms, size_mb):
    return {
        "Model": name,
        "Accuracy": round(accuracy_score(y_test_bin, pred), 4),
        "Precision": round(precision_score(y_test_bin, pred), 4),
        "Recall": round(recall_score(y_test_bin, pred), 4),
        "F1": round(f1_score(y_test_bin, pred), 4),
        "Train (s)": round(train_t, 2),
        "Latency (ms/sample)": round(infer_ms, 4),
        "Size (MB)": round(size_mb, 2),
    }

table = pd.DataFrame([
    row("Random Forest",    rf_bin,  rf_train_time,  rf_ms_per_sample, rf_size_mb),
    row("XGBoost",          xgb_bin, xgb_train_time, xgb_ms,           xgb_size_mb),
    row("Isolation Forest", iso_bin, iso_train_time, iso_ms,           iso_size_mb),
])
print("=== Binary comparison (Normal vs Attack) ===")
print(table.to_string(index=False))


X_train_normal = X_train[y_train == NORMAL_CLASS]
results = []
for c in [0.01, 0.05, 0.1, 0.2, 0.3, 0.5]:
    m = IsolationForest(n_estimators=100, contamination=c, random_state=42, n_jobs=-1)
    m.fit(X_train_normal)
    p = (m.predict(X_test) == -1).astype(int)
    results.append({
        "contamination": c,
        "Accuracy": round(accuracy_score(y_test_bin, p), 4),
        "Attack Precision": round(precision_score(y_test_bin, p), 4),
        "Attack Recall": round(recall_score(y_test_bin, p), 4),
        "F1": round(f1_score(y_test_bin, p), 4),
    })

print("\n=== Isolation Forest contamination sweep ===")
print(pd.DataFrame(results).to_string(index=False))
```

```python
import numpy as np
import pandas as pd


class_names = list(le.classes_)
NORMAL_CLASS = 7


severity_map = {
    'DDoS_UDP': 'High', 'DDoS_ICMP': 'High', 'DDoS_TCP': 'High', 'DDoS_HTTP': 'High',
    'Backdoor': 'Critical', 'MITM': 'Critical', 'Password': 'High',
    'Port_Scanning': 'Medium', 'Fingerprinting': 'Medium', 'Normal': 'None'
}

def generate_alerts(model, X_scaled_rows, row_index, original_df):
    """X_scaled_rows: scaled features; row_index: original df index labels"""
    t0 = time.time()
    proba = model.predict_proba(X_scaled_rows)
    pred = proba.argmax(axis=1)
    conf = proba.max(axis=1)
    latency_ms = (time.time() - t0) / len(pred) * 1000

    meta = original_df.loc[row_index, ['frame.time', 'ip.src_host', 'ip.dst_host']]
    alerts = pd.DataFrame({
        'timestamp': meta['frame.time'].values,
        'source_ip': meta['ip.src_host'].values,
        'destination_ip': meta['ip.dst_host'].values,
        'attack_type': [class_names[p] for p in pred],
        'confidence_score': np.round(conf, 4),
    })
    alerts['severity'] = alerts['attack_type'].map(severity_map)

   
    flagged = alerts[alerts['attack_type'] != 'Normal'].reset_index(drop=True)
    return flagged, latency_ms

flagged_alerts, lat = generate_alerts(xgb_model, X_test, y_test.index, df)

print(f"Alerts flagged: {len(flagged_alerts)} of {len(X_test)} test samples")
print(f"Alert generation latency: {lat:.4f} ms/sample")
print(flagged_alerts.head(10).to_string(index=False))

flagged_alerts.to_csv("flagged_alerts.csv", index=False)
print("\nSaved: flagged_alerts.csv  (pass to component 2)")
```
